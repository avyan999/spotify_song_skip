# Online Music Streaming Skip Rate Analysis

+ Extension: Analyze skip behavior → song popularity insights.
+ Add-ons: Bar charts, correlation.
+ Difficulty: Medium


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("spotify_skip_data.csv")
```

```python
df["skip_rate"] = df["skips"] / df["plays"]
```

```python
print("Dataset of Skip Rates for Spotify Songs")
df
```

```python
plt.figure(figsize=(10, 6))
plt.bar(df["song"], df["skip_rate"])
plt.xlabel("Song")
plt.ylabel("Skip Rate")
plt.title("Skip Rates for Spotify Songs")
plt.show()
```

```python
skip_popularity = df["skip_rate"].corr(df["popularity"])
skip_plays = df["skip_rate"].corr(df["plays"])
```

```python
print("Skip Popularity Correlation:", skip_popularity)
print("Skip Plays Correlation:", skip_plays)
```
